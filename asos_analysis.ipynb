```python
#Import dependencies

import pandas as pd
import matplotlib.pylab as plt
import seaborn as sns
```

```python
#Read data file.

df = pd.read_csv(r'C:\Users\h\products_asos.csv', on_bad_lines='skip')
#df['price'] = pd.to_numeric(df['price'], errors='coerce')

df.head(10)
```

```python
df.info()
```

```python
df['price'] = pd.to_numeric(df['price'], errors='coerce') # Convert price column to float type.
df = df.dropna(subset=['price']) # Drop any nan's
df.info()
```

```python
df.head(20)
```

```python
# Find the brands.

df['category'] = df['category'].astype(str) # Make sure category type is str.

def find_brand(text):
    if ' ' in text:
        try:
            return text.split(' ')[0].split(' ')[0]
        except:
            return 'Unknown'
    return 'Unknown'

df['brands'] = df['category'].apply(find_brand)   
```

```python
df.head(20)
```

```python

brand_counts = df['brands'].value_counts() # Number of times a brand is mentioned.
brand_limit = brand_counts[brand_counts>10].index # Limit the number of brand analysed.
df_clean = df[df['brands'].isin(brand_limit).copy()] # Clean dataframe.
print(df_clean['brands'].value_counts())
```

```python
#Analyse out of stock items.
def out_of_stock(size_str):
    if not isinstance(size_str, str):
        return 0, 0,0
    sizes = size_str.split(',')
    total_sizes = len(sizes)

    out_of_stock_count = size_str.count('Out of stock')
    out_of_stock_rate = out_of_stock_count/total_sizes if total_sizes>0 else 0

    return out_of_stock_count, out_of_stock_rate

metrics = df_clean['size'].apply(lambda x: out_of_stock(x))
df_clean['OutofStock_count'] = [x[0] for x in metrics]
df_clean['OutofStock_rate']  =[x[1] for x in metrics]

df_clean['Lost Revenue'] = df_clean['price']*df_clean['OutofStock_count']

cols = ['brands', 'name', 'price', 'OutofStock_count', 'Lost Revenue']
print(df_clean.sort_values(by='Lost Revenue', ascending=False).head(10) [cols])
```

```python
data = df_clean.groupby('brands').agg({
    'price': 'mean',
    'OutofStock_rate': 'mean',
    'Lost Revenue': 'sum',
    'name': 'count'
}).reset_index()

strategy = data[data['name']>10]

plt.figure(figsize=(12,8))
sns.scatterplot(data=strategy, x='price', y='OutofStock_rate', size='Lost Revenue', sizes=(50,500), alpha=0.7)

best_performers = strategy[
    (strategy['price']>40) &
    (strategy['OutofStock_rate']>0.35)]

for i in range(len(best_performers)):
    plt.text(
        best_performers.iloc[i]['price'],
        best_performers.iloc[i]['OutofStock_rate'],
        best_performers.iloc[i]['brands'])

plt.xlabel('Average Price')
plt.ylabel('Out of Stock Rate')
plt.axvline(x=40, color='r', ls='--')
plt.axhline(y=0.35, color='r', ls='--')
plt.title('Brand Analysis')
plt.show()
```
